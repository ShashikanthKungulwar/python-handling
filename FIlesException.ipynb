18. What is the difference between os.fork() and multiprocessing in Python?
    - os.fork() is a low-level Unix-only system call that duplicates the current process.
    - multiprocessing is a high-level, cross-platform Python module for creating and managing processes.
    - While os.fork() directly copies the parent’s memory (copy-on-write), multiprocessing starts fresh Python interpreters.
    - multiprocessing provides safe communication tools like Queue, Pipe, and Manager.
    - In general, multiprocessing is safer, portable, and preferred for parallel programming in Python

20. What is the difference between file.read() and file.readline() in Python?
    - file.read() reads the entire file (or a specified number of bytes) into a single string.

    - read() is useful for processing the whole file at once.
    - readline() is better when reading files line by line to save memory.
    - Calling readline() repeatedly continues from where it left off in the file.

```python
# 5 How would you catch and handle division by zero error in Python

try:
  5/0
except ZeroDivisionError as e:
  print("divided by zero check once")
```

```python
# 6 Write a Python program that logs an error message to a log file when a division by zero exception occurs?
import logging

logging.basicConfig(filename="error.log", level=logging.ERROR)

try:
    x = 10 / 0
except ZeroDivisionError as e:
    logging.error(f"Division by zero error occurred: {e}")

```

```python
#  7 How do you log information at different levels (INFO, ERROR, WARNING) in Python using the logging module?
import logging

# Configure logging
logging.basicConfig(filename="app.log", level=logging.DEBUG, format="%(asctime)s - %(levelname)s - %(message)s")

# Log messages at different levels
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```
